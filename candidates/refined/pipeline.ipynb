# Reproduce the single-model refined candidate
Run from this directory after installing requirements.txt. Supply the official competition test_images directory. Weights are included in weights/model.pt. No training or private Kaggle output is required.


```python
from pathlib import Path
import hashlib, json
provenance = json.loads(Path('provenance.json').read_text())
with Path('weights/model.pt').open('rb') as stream:
    assert hashlib.file_digest(stream, 'sha256').hexdigest() == provenance['checkpoint_sha256']
```

```python
import os, subprocess, sys
test_images = Path(os.environ['SOLAR_TEST_IMAGES'])
subprocess.run([sys.executable, '-m', 'src.submission', '--images', str(test_images),
    '--checkpoints', 'weights/model.pt', '--params', 'postprocess.json', '--device', 'cuda',
    '--tile', '512', '--tta', '--tta-batch-size', '4', '--output', 'reproduced.csv'], check=True)
```

```python
import cv2
from src.submission import validate_submission
shapes = {p.stem: cv2.imread(str(p), 0).shape for p in test_images.glob('*.jpeg')}
validate_submission('reproduced.csv', shapes)
```
